# MarketStream Engine — verified synthetic demo
This notebook calls the packaged production-oriented batch pipeline rather than copying implementation code. Synthetic data only. Run from the repository root after `pip install -e .`.


```python
from marketstream.config import load_config
from marketstream.processing.pipeline import execute_pipeline
cfg = load_config()
result = execute_pipeline(cfg, output_format='csv')
result.manifest['record_counts'], result.manifest['alignment']
```

```python
from pathlib import Path
import pandas as pd
from marketstream.storage.writers import resolve_current
out = resolve_current(result.destination, verify=True)
features = pd.read_csv(out / 'snapshot_features.csv')
bars = pd.read_csv(out / 'interval_bars.csv')
print('features', features.shape, 'bars', bars.shape)
display(features.head())
```

## Important interpretation
The last-known-trade categories are snapshot-based, not trade aggressor labels. Gap-fill observations are candidates from consecutive depth-five snapshots, not reconstructed orders. Chart quote-relative changes are not investment returns.
